# From forecasts to portfolio weights

Expected returns do not uniquely determine positions. This small long-only problem balances expected return against covariance risk, then compares the result with equal weighting.

## 1. Using a standard optimizer
Let SciPy enforce the long-only and fully invested constraints.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

mu = np.array([0.08, 0.06, 0.04])
cov = np.array([[.040, .012, .006], [.012, .025, .008], [.006, .008, .016]])
equal = np.ones(3) / 3

def solve(risk_aversion):
    utility = lambda w: -(w @ mu - risk_aversion * (w @ cov @ w) / 2)
    return minimize(utility, equal, bounds=[(0, 1)] * 3,
                    constraints={'type': 'eq', 'fun': lambda w: w.sum() - 1}).x

def summary(name, w):
    print(f'{name:12s} weights={w.round(3)}, return={w @ mu:.2%}, volatility={np.sqrt(w @ cov @ w):.2%}')

conservative = solve(8.0)
aggressive = solve(2.0)
summary('equal', equal); summary('conservative', conservative); summary('aggressive', aggressive)

x = np.arange(3); width = .25
plt.figure(figsize=(8, 4))
plt.bar(x - width, equal, width, label='equal')
plt.bar(x, conservative, width, label='more risk averse')
plt.bar(x + width, aggressive, width, label='less risk averse')
plt.xticks(x, ['Asset A', 'Asset B', 'Asset C']); plt.ylabel('weight'); plt.legend(); plt.show();
```

## 2. Build the optimizer from scratch
Take utility-gradient steps, then project negative weights to zero and renormalize to the simplex.

```python
def project_to_simplex(values):
    ordered = np.sort(values)[::-1]
    cumulative = np.cumsum(ordered)
    rho = np.where(ordered * np.arange(1, len(values) + 1) > cumulative - 1)[0][-1]
    threshold = (cumulative[rho] - 1) / (rho + 1)
    return np.maximum(values - threshold, 0)

def projected_gradient(risk_aversion, steps=3000, learning_rate=.05):
    w = np.ones(3) / 3
    for _ in range(steps):
        gradient = mu - risk_aversion * cov @ w
        w = project_to_simplex(w + learning_rate * gradient)
    return w

scratch = projected_gradient(3.0)
summary('from scratch', scratch)
```

## What to notice
The expected returns never change, yet the weights do. The risk preference and covariance matrix are part of the decision rule. Real portfolios add current holdings, turnover costs, liquidity, and many more constraints.
