# Financial text as model input

TF–IDF turns words and short phrases into sparse numerical features. A classifier can then learn which features distinguish the illustrative positive and negative documents.

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

docs = [
    'demand improved and margins expanded', 'cash flow strengthened with lower costs',
    'guidance increased after strong bookings', 'revenue growth exceeded expectations',
    'guidance fell as costs increased', 'demand weakened and margins contracted',
    'cash flow declined after a sales miss', 'revenue missed expectations and risk increased'
]
y = np.array([1, 1, 1, 1, 0, 0, 0, 0])
vectorizer = TfidfVectorizer(ngram_range=(1, 2))
X = vectorizer.fit_transform(docs)
model = LogisticRegression(C=2).fit(X, y)

tests = ['margins improved despite slower revenue', 'guidance fell and risk increased']
for text, probability in zip(tests, model.predict_proba(vectorizer.transform(tests))[:, 1]):
    print(f'{probability:.1%} positive  |  {text}')

terms = np.array(vectorizer.get_feature_names_out())
weights = model.coef_[0]
print('Most positive terms:', terms[np.argsort(weights)[-5:]][::-1])
print('Most negative terms:', terms[np.argsort(weights)[:5]])
```

## What to notice
The vocabulary and weights are fitted from the development documents. New phrases cannot contribute unless the representation knows them. In real research, every document also needs a version and an availability timestamp.
