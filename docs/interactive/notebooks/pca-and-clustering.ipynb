# PCA and clustering learn structure without a target

We will simulate two groups of correlated observations. PCA finds a direction that captures variation; K-means finds compact groups. Neither procedure sees a future return target.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

rng = np.random.default_rng(4)
factor = rng.normal(size=(240, 1))
group = np.repeat([0, 1], 120)
X = factor @ np.array([[1.0, 0.72]]) + np.c_[group * 1.4, group * -0.7] + rng.normal(0, .28, (240, 2))
Z = StandardScaler().fit_transform(X)
pca = PCA(n_components=2).fit(Z)
labels = KMeans(n_clusters=2, n_init=20, random_state=4).fit_predict(Z)

print(f'Variance captured by PC1: {pca.explained_variance_ratio_[0]:.1%}')
print('PC1 direction:', pca.components_[0].round(3))

plt.figure(figsize=(7, 6))
plt.scatter(Z[:, 0], Z[:, 1], c=labels, cmap='Set2', alpha=.7)
direction = pca.components_[0] * 3
plt.arrow(0, 0, direction[0], direction[1], width=.025, color='firebrick', length_includes_head=True)
plt.title('K-means groups and the first principal direction')
plt.xlabel('standardized feature 1'); plt.ylabel('standardized feature 2'); plt.axhline(0, color='.8'); plt.axvline(0, color='.8'); plt.show();
```

## What to notice
The first principal component is optimal for explaining variance in this sample. That does not make it optimal for forecasting. The clusters are similarly real as geometric groups without automatically being useful investment categories.
