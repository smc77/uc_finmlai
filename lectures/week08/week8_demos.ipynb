### Demo 1 — Three sizing schemes from one row of scores

> **Demonstration:** State what you expect before running the cell; afterward, explain which output supports or contradicts that expectation.

### One forecast, three ways to invest

**Question:** Does knowing the order tell us how much money to invest? `scores` contains 60 numbers, one per asset. A positive weight is a long position; a negative weight is a short position. Dollar-neutral means weights sum to zero; gross exposure one means their absolute values sum to one.

- `np.argsort(scores)` orders the assets. The quantile rule buys the top fifth and shorts the bottom fifth.
- The rank rule uses the centered position in that ordering. The magnitude rule uses the centered score itself.
- `normalize_dn` subtracts the mean and divides by total absolute weight. An all-zero input stays in cash.

**Predict → change → check:** multiply all scores by a positive constant, then use a nonlinear increasing transformation such as `np.tanh`. Which rules change? Check both `weights.sum()` and `np.abs(weights).sum()`. Equal ranking does not establish that the numerical score is a calibrated expected return. Calibration here means that forecast sizes correspond to the average outcomes they claim to predict.

```python
rng = np.random.default_rng(801)
K = 60
scores = rng.standard_normal(K)
ranks = np.argsort(np.argsort(scores))

def normalize_dn(w):
    """Dollar-neutral, gross exposure 1."""
    w = w - w.mean()
    gross = np.abs(w).sum()
    return w / gross if gross > 0 else np.zeros_like(w)

# Quantile L/S — top quintile +, bottom quintile -, else 0
q = K // 5
order = np.argsort(scores)
w_quant = np.zeros(K)
w_quant[order[-q:]] = 1.0; w_quant[order[:q]] = -1.0
w_quant = normalize_dn(w_quant)

# Rank-proportional
w_rank = normalize_dn(ranks - ranks.mean())

# Signal-scaled (z-scored)
z = (scores - scores.mean()) / scores.std()
w_sig = normalize_dn(z)

fig, ax = plt.subplots(figsize=(8, 3.5))
idx = np.argsort(scores)
ax.plot(scores[idx], w_quant[idx], "o-", label="quantile L/S", ms=3)
ax.plot(scores[idx], w_rank[idx], "s-", label="rank-prop", ms=3)
ax.plot(scores[idx], w_sig[idx], "^-", label="signal-scaled", ms=3)
ax.set_xlabel("score"); ax.set_ylabel("weight")
ax.set_title("Three sizing schemes — same forecast, three position profiles")
ax.legend(); ax.axhline(0, lw=0.5, color="k"); plt.show()
# Expected: one row of scores produces three different weight vectors. Sizing is a
#           decision, not a rescaling.
```

### Demo 2 — Volatility targeting on a regime-switching market

> **Demonstration:** State what you expect before running the cell; afterward, explain which output supports or contradicts that expectation.

### Scale exposure toward a risk target

**Question:** Can we take smaller positions when price swings grow? `sim.regime_switching` generates a 2,000-day return path that alternates between calm and turbulent states. This is the course simulation package, not downloaded market data.

`rolling(20).std().shift(1)` measures variability using only earlier returns. Divide the daily target, `0.10 / sqrt(252)`, by that estimate and cap exposure at five. Multiplying exposure by the subsequent return produces the strategy return. The first 20 days use exposure one because a full estimate is unavailable.

The three panels separate compounded wealth, annualized risk, and exposure; those quantities have different units. These are **gross** returns: fees, financing, and short-borrow charges are absent. This one-path, 20-day, five-times-cap example differs from the chapter's repeated 42-day, twice-cap experiment and its separate empirical overlay.

**Predict → change → check:** lower the cap to two. Which days change, and what happens to realized volatility? Stabilizing risk does not guarantee improving reward for risk. Inspect the path, not only the Sharpe ratio.

```python
r = np.asarray(sim.regime_switching(
    n=2000, sig_calm=0.006, sig_turb=0.022,
    p_stay_calm=0.985, p_stay_turb=0.96, seed=11
))

target_vol = 0.10 / np.sqrt(252)   # 10% annualized target
trailing = pd.Series(r).rolling(20).std().shift(1)   # past-only
lev = (target_vol / trailing).clip(0, 5).to_numpy()
lev = np.nan_to_num(lev, nan=1.0)

pnl_const = r
pnl_vt = lev * r

print(f"Constant Sharpe: {met.sharpe(pnl_const):+.2f}, MDD: {met.max_drawdown(pnl_const):.2%}")
print(f"Vol-targeted   : {met.sharpe(pnl_vt):+.2f}, MDD: {met.max_drawdown(pnl_vt):.2%}")

fig, axes = plt.subplots(1, 3, figsize=(12, 3.6))
axes[0].plot(np.r_[1, np.cumprod(1 + pnl_const)], label="constant size")
axes[0].plot(np.r_[1, np.cumprod(1 + pnl_vt)], label="vol targeted")
axes[0].set(xlabel="day", ylabel="wealth per initial dollar", title="Gross compounded wealth")
axes[0].legend(fontsize=8)
axes[1].plot(trailing * np.sqrt(252), color="grey", label="prior 20-day estimate")
axes[1].axhline(.10, color="tab:red", ls="--", label="annual target 10%")
axes[1].set(xlabel="day", ylabel="annualized volatility", title="Risk estimate uses past data")
axes[1].legend(fontsize=8)
axes[2].plot(lev, color="tab:blue", lw=.7)
axes[2].set(xlabel="day", ylabel="exposure multiple", title="Lower exposure when risk rises")
fig.suptitle("One path, 5x cap; no trading fees or financing charges")
fig.tight_layout(); plt.show()
```

### A target is not an executed trade

**Question:** Is following every small forecast change worth its cost? The same fixed score path supplies targets for every policy. A no-trade band is a tolerance for leaving the existing holdings alone. Here it is the **full-book L1 gap**, the sum of absolute weight differences across all assets; it is not a separate band for each asset.

`execute_policy(targets, subsequent_returns, cost_bps, threshold)` follows a row in order:

1. Compare the target with current holdings, including drift since the last return.
2. Enter the target from cash, or rebalance when the whole-book gap exceeds the threshold; otherwise retain the holdings.
3. Compute traded dollars as `abs(actual - before).sum()`. A move from −0.5 to +0.5 counts as one dollar per dollar of wealth.
4. Earn `actual @ realized` and subtract `cost_bps / 10000 * traded` for that same interval.
5. Carry forward `actual * (1 + realized) / (1 + net_return)`. The denominator includes the fee; residual cash has zero interest. Financing, market impact, and short-borrow charges are omitted.

Targets at row `t` are paired with returns at `t+1`. The first 1,099 such pairs select among thresholds 0, 0.4, 0.8, 1.2, and 1.6; the next 1,100 evaluate the frozen choice. Each block starts from cash and pays entry. All test curves are shown for teaching; choosing again from them would consume the test as development data.

**Predict → change → check:** inspect the selected policy's nonzero-trade count, tracking error, net exposure, and gross exposure. A large threshold may almost stop trading. Neutral targets do not keep drifting holdings neutral. Trace the first two days by hand before interpreting Sharpe. The preceding raw-versus-smoothed comparison changes the forecasts and is a different question from this fixed-forecast threshold experiment.

```python
rng = np.random.default_rng(803)
T, K = 2200, 30
score_path = rng.standard_normal((T, K)) * 0.5 + rng.standard_normal((1, K)) * 0.3
fwd_ret = rng.standard_normal((T, K)) * 0.01
fwd_ret[1:] += 0.0005 * score_path[:-1]  # score at t predicts return at t+1

def weights_from_score(s):
    s = s - s.mean(1, keepdims=True)
    gross = np.abs(s).sum(1, keepdims=True)
    return np.divide(s, gross, out=np.zeros_like(s), where=gross > 0)

w_raw = weights_from_score(score_path)
# EWMA smoothed
alpha = 0.2
s_smooth = np.zeros_like(score_path); s_smooth[0] = score_path[0]
for t in range(1, T):
    s_smooth[t] = alpha * score_path[t] + (1 - alpha) * s_smooth[t-1]
w_smooth = weights_from_score(s_smooth)

# Each target at t earns the subsequent return at t+1. The trade that
# establishes that holding is charged in the same row, including entry from cash.
def execute_policy(targets, subsequent_returns, cost_bps=2.0, threshold=0.0):
    """Full-book L1 inaction rule; weights measured against opening net wealth.

    Rebalance the whole book if target-minus-drifted-holding L1 distance exceeds
    the threshold. Otherwise keep the actual risky dollar holdings. Cash absorbs
    the linear fee; borrowing and cash earn zero here. Targets are neutral/gross
    one; realized holdings may drift between trades. No financing/impact model.
    """
    targets = np.asarray(targets, float)
    returns = np.asarray(subsequent_returns, float)
    if targets.shape != returns.shape:
        raise ValueError('Each target needs one subsequent-return row')
    before = np.zeros(targets.shape[1])
    held, gross, net, turnover, drifted = [], [], [], [], []
    for t, (target_weight, realized) in enumerate(zip(targets, returns)):
        gap = np.abs(target_weight - before).sum()
        actual = target_weight.copy() if t == 0 or gap > threshold else before.copy()
        traded = np.abs(actual - before).sum()
        gain = float(actual @ realized)
        after_fee = gain - cost_bps * 1e-4 * traded
        if 1 + after_fee <= 0:
            raise ValueError('The simulated policy exhausted its equity')
        held.append(actual); gross.append(gain); net.append(after_fee); turnover.append(traded)
        before = actual * (1 + realized) / (1 + after_fee)
        drifted.append(before.copy())
    return {key: np.asarray(value) for key, value in
            [('held', held), ('gross', gross), ('net', net),
             ('turnover', turnover), ('post_return', drifted)]}


def strat_pnl(w, ret, cost_bps):
    return execute_policy(w[:-1], ret[1:], cost_bps=cost_bps)['net']

bps_grid = np.linspace(0, 8, 17)
net_raw = [met.sharpe(strat_pnl(w_raw, fwd_ret, c)) for c in bps_grid]
net_smooth = [met.sharpe(strat_pnl(w_smooth, fwd_ret, c)) for c in bps_grid]
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(bps_grid, net_raw, 'o-', label='raw scores')
ax.plot(bps_grid, net_smooth, 's-', label='smoothed scores')
ax.axhline(0, color='k', lw=0.5)
ax.set(xlabel='cost (bps per full-L1 dollar traded)', ylabel='net Sharpe',
       title='Smoothing changes both the forecast and its trading cost')
ax.legend(); fig.tight_layout()
save_lecture_figure(fig, 'demo3_cost_inversion.png'); plt.show()

# Main task: one unchanged target path, five implementation policies. Each
# block starts from cash so every candidate pays its own entry cost.
policy_cut = 1100
validation_count = policy_cut - 1
aligned_target, aligned_return = w_raw[:-1], fwd_ret[1:]
bands = [0.0, 0.4, 0.8, 1.2, 1.6]
validation_results = {
    band: execute_policy(aligned_target[:validation_count], aligned_return[:validation_count],
                         threshold=band) for band in bands
}
selected_band = max(bands, key=lambda band: met.sharpe(validation_results[band]['net']))
# The choice is frozen above, before any test results are computed. The test
# curve below is a diagnostic sweep, never a second opportunity to choose.
band_rows, band_paths = [], {}
for band in bands:
    test = execute_policy(aligned_target[validation_count:], aligned_return[validation_count:],
                          threshold=band)
    band_paths[band] = test
    band_rows.append({
        'threshold': band,
        'validation_net_Sharpe': met.sharpe(validation_results[band]['net']),
        'test_net_Sharpe': met.sharpe(test['net']),
        'test_tracking_error': np.mean(np.abs(test['held'] - aligned_target[validation_count:])),
        'test_turnover': test['turnover'].mean(),
        'test_mean_gross_exposure': np.abs(test['held']).sum(axis=1).mean(),
        'test_max_abs_net_exposure': np.abs(test['held'].sum(axis=1)).max(),
    })
band_frame = pd.DataFrame(band_rows)
print('Validation selects; test curves are diagnostic only:')
print(band_frame.round(3).to_string(index=False))
print(f'Selected full-book L1 threshold: {selected_band:.1f}')
fig, axes = plt.subplots(1, 4, figsize=(13, 3.4))
for ax, column, title in zip(axes,
    ['validation_net_Sharpe', 'test_net_Sharpe', 'test_turnover', 'test_tracking_error'],
    ['Select on validation', 'Frozen choice on test', 'Test trading', 'Test target gap']):
    ax.plot(band_frame['threshold'], band_frame[column], 'o-', lw=2)
    chosen = band_frame[band_frame['threshold'] == selected_band].iloc[0]
    ax.scatter([selected_band], [chosen[column]], s=75, marker='*', color='tab:red', zorder=4)
    ax.set(xlabel='full-book L1 threshold', title=title); ax.grid(alpha=.2)
for ax in axes[:2]: ax.set_ylabel('net Sharpe at 2 bps')
axes[2].set_ylabel('full-L1 dollars / day')
axes[3].set_ylabel('mean absolute weight gap')
fig.suptitle('Demo 3: trade from actual holdings, then earn the next return')
fig.tight_layout(rect=(0, 0, 1, .91))
save_lecture_figure(fig, 'demo3_no_trade_bands.png'); plt.show()
```

### Two nearly identical assets, different decisions

**Question:** Why can a small estimate change move a portfolio so far? The first 60 returns estimate two means and a covariance matrix. The `lagrange` and `unconstrained` calculations solve the fully invested two-asset quadratic utility problem. The unconstrained result can short one asset and borrow to hold the other. Clipping it to `[0, 1]` is the exact long-only solution for this particular two-asset problem.

Equal weights ignore both estimated means and covariance. Inverse-volatility weights use marginal risk estimates and normalize to sum one. Read the moving Asset A weight beside the nearly identical price paths.

**Predict → change → check:** increase the estimation window and compare weight movement. The reported absolute change in Asset A's target is half the two-asset full-L1 target change. It excludes entry and holdings drift, so it is **not** the actual traded-dollar cost ledger used in Demo 3.

```python
pair_rng = np.random.default_rng(808)
pair_dates = pd.bdate_range("2024-01-02", periods=520)
pair_common = pair_rng.normal(0.00025, 0.009, len(pair_dates))
pair_returns = pd.DataFrame(
    {
        "Asset A": pair_common + pair_rng.normal(0, 0.0013, len(pair_dates)),
        "Asset B": pair_common + pair_rng.normal(0, 0.0013, len(pair_dates)),
    },
    index=pair_dates,
)
pair_window = 60
pair_gamma = 100.0
pair_rows = []
for endpoint in range(pair_window, len(pair_returns)):
    history = pair_returns.iloc[endpoint - pair_window:endpoint]
    mean = history.mean().to_numpy() * 252
    covariance = history.cov().to_numpy() * 252
    inverse = np.linalg.inv(covariance)
    ones = np.ones(2)
    # Fully invested unconstrained quadratic-utility solution. The Lagrange
    # multiplier imposes w_A + w_B = 1 while permitting the long/short pair.
    lagrange = (ones @ inverse @ mean - pair_gamma) / (ones @ inverse @ ones)
    unconstrained = inverse @ (mean - lagrange * ones) / pair_gamma
    inverse_vol = 1 / np.sqrt(np.diag(covariance))
    inverse_vol /= inverse_vol.sum()
    pair_rows.append(
        {
            "date": pair_returns.index[endpoint],
            "correlation": history.corr().iloc[0, 1],
            "mean spread": mean[0] - mean[1],
            "plug-in": unconstrained[0],
            "long-only plug-in": np.clip(unconstrained[0], 0, 1),
            "equal weight": 0.5,
            "inverse volatility": inverse_vol[0],
        }
    )
pair_rolling = pd.DataFrame(pair_rows).set_index("date")

fig, axes = plt.subplots(1, 3, figsize=(11, 3.2))
(1 + pair_returns).cumprod().plot(ax=axes[0], lw=1.4)
axes[0].set_title("The return paths look redundant")
axes[0].set_ylabel("growth of $1")
axes[0].legend(frameon=False, fontsize=8)
axes[1].plot(pair_rolling.index, pair_rolling["correlation"], color="#3b6fb6")
axes[1].axhline(pair_rolling["correlation"].median(), color="0.25", ls=":", lw=1)
axes[1].set_ylim(0.94, 1.001)
axes[1].set_title("Rolling correlation stays near one")
axes[1].set_ylabel("60-day correlation")
axes[2].plot(pair_rolling.index, pair_rolling["plug-in"], color="#b23a48", lw=1.2)
axes[2].axhline(0.5, color="0.25", ls=":", lw=1)
axes[2].set_title("Yet the optimized A weight swings")
axes[2].set_ylabel("fully invested weight in A")
for ax in axes:
    ax.grid(alpha=0.18)
for ax in axes[1:]:
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
fig.tight_layout()
save_lecture_figure(fig, "pair_mean_variance_instability.png")
plt.show()

method_columns = ["long-only plug-in", "equal weight", "inverse volatility"]
method_colors = ["#b23a48", "#4c956c", "#d89b2b"]
turnover = {}
fig, axes = plt.subplots(1, 2, figsize=(10, 3.25), gridspec_kw={"width_ratios": [2.2, 1]})
for method, color in zip(method_columns, method_colors):
    axes[0].plot(pair_rolling.index, pair_rolling[method], label=method, color=color, lw=1.35)
    turnover[method] = pair_rolling[method].diff().abs().dropna().mean()
axes[0].set_ylim(-0.04, 1.04)
axes[0].set_title("How much capital goes to Asset A?")
axes[0].set_ylabel("long-only weight")
axes[0].legend(frameon=False, fontsize=8, ncol=3, loc="lower center")
axes[1].bar(range(3), [turnover[name] for name in method_columns], color=method_colors)
axes[1].set_xticks(range(3), ["plug-in", "1/N", "inverse vol"])
axes[1].set_title("Daily target turnover")
axes[1].set_ylabel("mean absolute change")
for index, method in enumerate(method_columns):
    axes[1].text(index, turnover[method] + 0.001, f"{turnover[method]:.3f}", ha="center", fontsize=8)
for ax in axes:
    ax.grid(alpha=0.18, axis="y")
fig.tight_layout()
save_lecture_figure(fig, "pair_low_estimation_allocators.png")
plt.show()
print("Median rolling correlation:", round(pair_rolling["correlation"].median(), 3))
print("Unconstrained A-weight range:", tuple(pair_rolling["plug-in"].agg(["min", "max"]).round(2)))
print("Mean absolute daily Asset A target change:", {name: round(value, 4) for name, value in turnover.items()})
# Expected: two nearly redundant return paths generate a volatile plug-in
# allocation. Equal weight ignores the noisy distinction; inverse volatility
# estimates only marginal risks, so both are much more stable in this example.
```

### Compare allocations under one mandate

**Question:** Which rule produces low true risk when risk must be estimated? `one_sample` creates a planted factor covariance and draws 120 returns for 20 assets. `np.cov` estimates covariance; `LedoitWolf().fit(returns).covariance_` shrinks the estimate toward a scaled identity. Shrinkage changes the risk estimate, not expected returns.

`long_only_gmv` minimizes `weight @ covariance @ weight`. Its equality constraint makes weights sum to one; its bounds prevent shorting. SLSQP is the numerical solver, and a failed solve raises an error. Equal weight and inverse volatility share those investment constraints but use different rules; they do not optimize the same estimated objective.

The oracle solves the same constrained problem using the simulator's true covariance. Each candidate's true variance is divided by oracle variance, so one is the feasible floor. The 120 repetitions redraw estimation samples. Boxplots show sample variation, not 120 market histories.

**Predict → change → check:** reduce the sample length toward the asset count. Compare the full distributions, not just the best median. Shrinkage need not win every planted population.

```python
def long_only_gmv(covariance):
    """Long-only, fully invested global-minimum-variance portfolio."""
    n_assets = covariance.shape[0]
    result = minimize(
        lambda weight: float(weight @ covariance @ weight),
        np.repeat(1 / n_assets, n_assets),
        jac=lambda weight: 2 * covariance @ weight,
        method="SLSQP",
        bounds=[(0.0, 1.0)] * n_assets,
        constraints={
            "type": "eq",
            "fun": lambda weight: weight.sum() - 1.0,
            "jac": lambda weight: np.ones_like(weight),
        },
        options={"ftol": 1e-10, "maxiter": 250},
    )
    if not result.success:
        raise RuntimeError(result.message)
    return result.x


def one_sample(K=20, T=120, seed=None):
    g = np.random.default_rng(seed)
    vol = np.linspace(0.012, 0.026, K)
    loadings = np.column_stack([
        np.linspace(0.2, 1.0, K),
        np.sin(np.linspace(0, 2 * np.pi, K)),
    ])
    factor_cov = np.diag([0.00010, 0.00004])
    cov = loadings @ factor_cov @ loadings.T + np.diag(0.45 * vol ** 2)
    returns = g.multivariate_normal(np.zeros(K), cov, size=T)
    sample_cov = np.cov(returns, rowvar=False)
    shrink_cov = LedoitWolf().fit(returns).covariance_

    oracle = long_only_gmv(cov)
    oracle_variance = float(oracle @ cov @ oracle)
    w_eq = np.ones(K) / K
    w_iv = 1 / np.sqrt(np.diag(sample_cov)); w_iv /= w_iv.sum()
    weights = {
        "Equal weight": w_eq,
        "Inverse vol": w_iv,
        "Sample GMV": long_only_gmv(sample_cov),
        "LW GMV": long_only_gmv(shrink_cov),
    }
    return {
        name: float(weight @ cov @ weight) / oracle_variance
        for name, weight in weights.items()
    }

results = [one_sample(seed=i) for i in range(120)]
df = pd.DataFrame(results)
print("Median population variance relative to oracle GMV:")
print(df.median().round(3))

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.boxplot(
    [df[c].dropna() for c in df.columns],
    tick_labels=df.columns,
    showmeans=True,
)
ax.set_title("One mandate: long-only, fully invested, minimum variance")
ax.axhline(1, color="k", lw=0.8, ls=":", label="oracle GMV")
ax.set_ylabel("population variance / oracle variance")
ax.legend(frameon=False)
fig.tight_layout()
save_lecture_figure(fig, "demo5_robust_allocations.png")
plt.show()
# Shrinkage does not win every population: inspect the common-mandate risk ratios.
```

### A better-conditioned matrix is an intermediate result

**Question:** Does changing the risk estimate improve the portfolio it produces? Forty assets have three shared factors plus independent risk; only 60 returns estimate their covariance. Eigenvalues describe variation along different portfolio directions. The condition number divides the largest by the smallest: a large ratio signals numerically sensitive directions.

`LedoitWolf().fit(X).covariance_` produces the shrinkage estimate. The left log-scale plot compares eigenvalues. The right panel compares relative matrix error and **excess** true GMV variance over the oracle: `risk_ratio - 1`. A bar of 0.183 means 18.3% more variance than the oracle, not a variance ratio of 0.183.

```python
rng = np.random.default_rng(806)
N, T = 40, 60
factor_loadings = rng.normal(0, 0.006, size=(N, 3))
true_covariance = factor_loadings @ factor_loadings.T + np.diag(
    rng.uniform(0.00004, 0.00012, N)
)
X = rng.multivariate_normal(np.zeros(N), true_covariance, size=T)
S = np.cov(X.T)
lw = LedoitWolf().fit(X).covariance_

eS = np.sort(np.linalg.eigvalsh(S))[::-1]
eLW = np.sort(np.linalg.eigvalsh(lw))[::-1]
print(f"Condition number S : {eS[0] / eS[-1]:.0f}")
print(f"Condition number LW: {eLW[0] / eLW[-1]:.0f}")
errors = {
    "sample": np.linalg.norm(S - true_covariance, "fro") / np.linalg.norm(true_covariance, "fro"),
    "Ledoit-Wolf": np.linalg.norm(lw - true_covariance, "fro") / np.linalg.norm(true_covariance, "fro"),
}
oracle_gmv = long_only_gmv(true_covariance)
oracle_risk = float(oracle_gmv @ true_covariance @ oracle_gmv)
risk_ratios = {
    "sample": float(long_only_gmv(S) @ true_covariance @ long_only_gmv(S)) / oracle_risk,
    "Ledoit-Wolf": float(long_only_gmv(lw) @ true_covariance @ long_only_gmv(lw)) / oracle_risk,
}
print("Relative covariance error:", {k: round(v, 3) for k, v in errors.items()})
print("Population GMV risk / oracle:", {k: round(v, 3) for k, v in risk_ratios.items()})

fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].semilogy(eS, "o-", label="sample")
axes[0].semilogy(eLW, "s-", label="Ledoit-Wolf")
axes[0].set_title("Estimated covariance spectrum")
axes[0].set_xlabel("ordered covariance direction")
axes[0].set_ylabel("eigenvalue")
axes[0].legend()
labels = ["sample", "Ledoit-Wolf"]
x = np.arange(len(labels)); width = .34
axes[1].bar(x - width/2, [errors[k] for k in labels], width, label="covariance error")
axes[1].bar(x + width/2, [risk_ratios[k] - 1 for k in labels], width, label="excess GMV risk")
axes[1].set_xticks(x, labels)
axes[1].set_title("Known-truth consequences")
axes[1].set_ylabel("relative loss")
axes[1].axhline(0, color="k", lw=.6)
axes[1].legend(frameon=False)
fig.tight_layout()
save_lecture_figure(fig, "demo6_ledoit_wolf_spectrum.png")
plt.show()
# Expected: shrinkage improves conditioning and reduces known-truth covariance
#           loss and portfolio risk without collapsing a real factor structure.
```

### Check a derivative two independent ways

**Question:** Did we correctly calculate how a weight affects prediction error? One input passes through a tanh hidden unit and a linear output. The four parameters are the two weights and two biases. Loss is half the squared error. `scalar_loss_and_gradient` applies the chain rule backward; the numerical check calls the loss at `theta + h` and `theta - h` for each parameter and divides by `2*h`.

This is the exact parameter point worked on the lecture's numerical slide. The chapter figure uses a different parameter point to explain the same checking method. The log-log chart reports the largest discrepancy over all four parameters for each step size.

**Predict → change → check:** make `h` extremely small. Does agreement keep improving? Large steps cease to be local; tiny ones lose floating-point precision. This checks code locally, not generalization. An optional visual explanation is [3Blue1Brown's backpropagation lesson](https://www.3blue1brown.com/lessons/backpropagation/): predict which connection changes the loss most, then trace the derivative path. Use the local numerical example to check the intuition.

```python
parameters = np.array([0.8, -0.3, 1.2, 0.1])  # w1, b1, w2, b2
x_value, target_value = 0.7, -0.4

def scalar_loss_and_gradient(theta):
    w1, b1, w2, b2 = theta
    hidden = np.tanh(w1 * x_value + b1)
    prediction = w2 * hidden + b2
    error = prediction - target_value
    loss = 0.5 * error ** 2
    hidden_grad = error * w2 * (1 - hidden ** 2)
    gradient = np.array([hidden_grad * x_value, hidden_grad, error * hidden, error])
    return loss, gradient

loss, analytic = scalar_loss_and_gradient(parameters)
steps = np.logspace(-14, -1, 28)
errors = []
for step in steps:
    numeric = np.zeros_like(parameters)
    for j in range(len(parameters)):
        up, down = parameters.copy(), parameters.copy()
        up[j] += step
        down[j] -= step
        numeric[j] = (
            scalar_loss_and_gradient(up)[0] - scalar_loss_and_gradient(down)[0]
        ) / (2 * step)
    errors.append(np.max(np.abs(numeric - analytic)))

best = int(np.argmin(errors))
print("Analytic gradient:", np.round(analytic, 6))
print(f"Best finite-difference step={steps[best]:.1e}; max error={errors[best]:.2e}")
print("Very small steps lose precision; large steps cease to be local.")

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.loglog(steps, errors, "o-")
ax.axvline(steps[best], color="tab:red", ls="--", lw=1)
ax.set_xlabel("central-difference step")
ax.set_ylabel("maximum absolute gradient error")
ax.set_title("A gradient check fails at both numerical extremes")
plt.tight_layout()
save_lecture_figure(fig, "demo7_gradient_check.png")
plt.show()
```

### Representation and training are separate questions

**Question:** Can the fitting algorithm find a curve the network can represent? `MLPRegressor(hidden_layer_sizes=(h,), activation='tanh')` creates one hidden layer of width `h`. `fit(X, y)` adjusts weights using Adam; `predict(X)` applies the fitted network to the same 400 noiseless grid points. Compare widths 2, 5, 20, and 50.

This is **training-grid approximation**, not held-out prediction. The chapter's theorem slide instead constructs ReLU interpolation weights directly from knot slopes without an optimizer. That construction demonstrates representability; this fitted tanh example also depends on initialization, stopping, and optimization. Wider is not guaranteed to fit better in one run. Suppressed convergence warnings do not establish convergence; inspect `nn.n_iter_` and `nn.loss_curve_`.

**Predict → change → check:** keep width fixed and change `random_state` or training iterations. Which differences come from capacity and which from fitting? Evaluate on a fresh grid before making an interpolation claim, and outside the original range before making an extrapolation claim.

### Demo 10 — Capacity is relative to the data budget

> **Break cue:** Deck B, after recording segment S5. Read the task on the preceding slide, predict what the output should show, then run the cell. ### More data, more width, or a different domain?

**Question:** When does flexibility recover a real curve rather than noise? This is a fixed random-feature network. `tanh(x * hidden_w + hidden_b)` creates 4, 12, or 32 hidden features. Those hidden weights never train. A regularized linear solve fits only the output coefficients, leaving the intercept unpenalized.

Training sets contain 40, 120, or 300 inputs in `[-2, 2]`, with noise added to the known target. Ten independent training draws per cell reveal sample variation; hidden features are fixed within each width. Different widths have different fixed feature draws, so width and that draw are not fully isolated.

The three heatmaps report median error on noisy training labels, the known curve inside the range, and the known curve in two outside intervals: `[-2.7, -2.05]` and `[2.05, 2.7]`. Each heatmap uses its own color scale. The extra charts show the 10th–90th-percentile sample spread and the median paired ratio of shifted to in-domain error. These are descriptive spreads, not confidence intervals.

**Predict → change → check:** distinguish more examples inside the same range from examples covering a wider range. No validation procedure selects a width here: these are diagnostic comparisons. If you choose from the plotted results, obtain a separate final evaluation. A successful curve inside familiar inputs does not establish performance outside them.

```python
widths = [4, 12, 32]
sample_sizes = [40, 120, 300]
x_grid = np.linspace(-2, 2, 600)
x_shift = np.r_[np.linspace(-2.7, -2.05, 200), np.linspace(2.05, 2.7, 200)]
n_repeats = 10
train_mse = np.zeros((len(sample_sizes), len(widths), n_repeats))
truth_mse = np.zeros_like(train_mse)
shift_mse = np.zeros_like(train_mse)

# This is a one-hidden-layer tanh network with frozen hidden weights and a
# fitted linear readout. The construction isolates width and data budget from
# optimizer luck while remaining a genuine nonlinear network.
for i, n in enumerate(sample_sizes):
    for repeat in range(n_repeats):
        g = np.random.default_rng(808 + 100 * i + repeat)
        x_train = np.sort(g.uniform(-2, 2, n))
        y_train = target(x_train) + 0.30 * g.standard_normal(n)
        for j, width in enumerate(widths):
            hidden_rng = np.random.default_rng(700 + j)
            hidden_w = hidden_rng.normal(0, 2, width)
            hidden_b = hidden_rng.uniform(-2, 2, width)

            def design(values):
                hidden = np.tanh(values[:, None] * hidden_w + hidden_b)
                return np.column_stack([np.ones(len(values)), hidden])

            phi_train = design(x_train)
            penalty = np.diag(np.r_[0.0, np.full(width, 1e-8)])
            readout = np.linalg.solve(
                phi_train.T @ phi_train + penalty, phi_train.T @ y_train
            )
            train_mse[i, j, repeat] = np.mean((y_train - phi_train @ readout) ** 2)
            truth_mse[i, j, repeat] = np.mean((target(x_grid) - design(x_grid) @ readout) ** 2)
            shift_mse[i, j, repeat] = np.mean((target(x_shift) - design(x_shift) @ readout) ** 2)

surfaces = [np.median(values, axis=2) for values in (train_mse, truth_mse, shift_mse)]
print("Median grid MSE against known truth over ten training-sample draws:")
print(pd.DataFrame(surfaces[1], index=sample_sizes, columns=widths).round(3))

fig, axes = plt.subplots(1, 3, figsize=(12, 3.7))
for ax, values, title in zip(
    axes, surfaces,
    ["Fit to noisy labels", "Fresh in-domain truth", "Shifted-domain truth"],
):
    im = ax.imshow(values, cmap="viridis_r", aspect="auto")
    ax.set_xticks(range(len(widths)), widths)
    ax.set_yticks(range(len(sample_sizes)), sample_sizes)
    ax.set_xlabel("hidden width")
    ax.set_ylabel("training observations")
    ax.set_title(title)
    for i in range(len(sample_sizes)):
        for j in range(len(widths)):
            ax.text(j, i, f"{values[i, j]:.2f}", ha="center", va="center",
                    color="white" if values[i, j] > np.median(values) else "black")
fig.suptitle("Demo 10: useful capacity depends on data budget and evaluation domain\n"
             "Each panel has its own color scale; cell values are median MSE")
fig.subplots_adjust(top=.78, right=.98, wspace=.34)
save_lecture_figure(fig, "demo10_capacity_data_budget.png")
plt.show()
# Expected: the best width depends on the data budget: width 12 wins at n=40, and
#           width 32 only becomes competitive as n grows.


# Revisit the same experiment with new questions: sample variation, then domain shift.
fig, ax = plt.subplots(figsize=(7.8, 3.8))
for j, width in enumerate(widths):
    lo, median, hi = np.quantile(truth_mse[:, j, :], [.1, .5, .9], axis=1)
    ax.plot(sample_sizes, median, 'o-', label=f'width {width}')
    ax.fill_between(sample_sizes, lo, hi, alpha=.15)
ax.set(xlabel='training observations', ylabel='in-domain truth MSE', yscale='log',
       title='More data changes both error and sample-to-sample variation')
ax.legend(); fig.tight_layout()
save_lecture_figure(fig, 'demo10_capacity_uncertainty.png'); plt.show()
fig, ax = plt.subplots(figsize=(7.8, 3.8))
for j, width in enumerate(widths):
    ratio = shift_mse[:, j, :] / truth_mse[:, j, :]
    lo, median, hi = np.quantile(ratio, [.1, .5, .9], axis=1)
    ax.plot(sample_sizes, median, 'o-', label=f'width {width}')
    ax.fill_between(sample_sizes, lo, hi, alpha=.15)
ax.axhline(1, color='gray', ls='--')
ax.set(xlabel='training observations', ylabel='shifted / in-domain truth MSE', yscale='log',
       title='Error can rise sharply outside the observed input range')
ax.legend(); fig.tight_layout()
save_lecture_figure(fig, 'demo10_domain_shift.png'); plt.show()
```

### Match a model to the planted pattern

**Question:** Which fitted procedure recovers this hidden tabular function? `make_panel` creates 1,800 independent rows with 20 features and threshold interactions, then adds noisy outcomes. The first 60% train; the last 40% test. `StandardScaler().fit` learns means and scales from training rows, then `transform` applies those values to test rows.

Ridge fits a penalized linear prediction. `GradientBoostingRegressor.fit` adds trees that repair residual errors. `MLPRegressor.fit` learns two hidden layers, using a random internal validation subset for stopping. That random split is appropriate for these independent toy rows; it would not implement a forward test of persistent market data. The fitting budgets and stopping rules differ across models.

Five repetitions report truth MSE and noisy-outcome rank IC. Truth MSE compares forecasts with the simulator's hidden mean. Rank IC is the Spearman correlation between forecast and outcome rankings. Their information content differs because realized outcomes contain noise.

**Predict → change → check:** replace the threshold target with a linear target, keeping the noise and splits fixed. This is a conditional comparison, not a claim that one algorithm is always best or a complete tuning-budget study.

```python
def make_panel(T=2000, K=20, seed=0):
    g = np.random.default_rng(seed)
    X = g.standard_normal((T, K))
    # Local thresholds and interactions are a favorable known-truth case for trees.
    signal = (
        0.30 * ((X[:, 0] > 0.4) & (X[:, 1] < -0.2))
        - 0.24 * ((X[:, 2] < -0.5) & (X[:, 3] > 0.0))
        + 0.18 * np.where(X[:, 4] > 0.8, X[:, 5], 0.0)
    )
    y = signal + 0.65 * g.standard_normal(T)
    return X, y, signal

comparison = []
for seed in range(5):
    X, y, truth = make_panel(T=1800, seed=seed)
    ix = int(0.6 * len(y))
    X_tr, X_te = X[:ix], X[ix:]
    y_tr, y_te = y[:ix], y[ix:]
    truth_te = truth[ix:]
    scaler = StandardScaler().fit(X_tr)
    Xs_tr, Xs_te = scaler.transform(X_tr), scaler.transform(X_te)
    models = {
        "Ridge": Ridge(alpha=1.0),
        "Boosting": GradientBoostingRegressor(
            n_estimators=100, max_depth=2, learning_rate=0.05, random_state=seed
        ),
        "MLP": MLPRegressor(
            hidden_layer_sizes=(16, 16), max_iter=250, early_stopping=True,
            validation_fraction=0.2, n_iter_no_change=12, random_state=seed
        ),
    }
    for name, model in models.items():
        model.fit(Xs_tr, y_tr)
        pred = model.predict(Xs_te)
        comparison.append({
            "seed": seed,
            "model": name,
            "truth_MSE": np.mean((pred - truth_te) ** 2),
            "noisy_rank_IC": met.rank_ic(pred, y_te),
        })

comparison = pd.DataFrame(comparison)
summary = comparison.groupby("model")[["truth_MSE", "noisy_rank_IC"]].median()
print("Median over five out-of-sample simulations:")
print(summary.round(3))

fig, axes = plt.subplots(1, 2, figsize=(9.5, 3.6))
order = ["Ridge", "Boosting", "MLP"]
axes[0].bar(order, summary.loc[order, "truth_MSE"])
axes[0].set_title("Known-truth error")
axes[0].set_ylabel("MSE against conditional mean")
axes[1].bar(order, summary.loc[order, "noisy_rank_IC"])
axes[1].set_title("Noisy-label rank IC")
axes[1].axhline(0, color="k", lw=0.5)
fig.suptitle("The same model comparison under two evaluation targets")
plt.tight_layout(rect=(0, 0, 1, .91))
save_lecture_figure(fig, "demo11_tabular_bakeoff.png")
plt.show()
# Expected: boosting recovers the conditional mean better than the MLP (truth MSE ~0.03
#           vs ~0.06) while the noisy rank IC hides most of that gap.
```

### Save the chosen network, not just its epoch number

**Question:** When should we stop learning from these training examples? A new 900-row sample is divided in order: 300 train, 300 validation, 300 test. The population is independent here, but the ordered split models the forward protocol we need with financial time series. Scaling is fitted only on the training rows.

`fit_forward_mlp` repeatedly calls `partial_fit` for one epoch, preserving Adam's internal optimizer state. Training loss describes fit to seen examples; validation loss chooses the best snapshot. `deepcopy(model)` saves a separate model, so later updates cannot overwrite it. The helper never accepts test data. After all 120 candidate epochs are fixed, the final 300 rows compare the saved checkpoint with the terminal model.

**Predict → change → check:** mark the validation minimum before viewing test loss. Inspect `best_epoch`, the two test errors, and the actual restored predictions. This fixture selects epoch 5; the earlier chapter figure uses a separate 320-epoch experiment. Neither value is a general stopping rule. A smaller validation minimum is selected evidence, not an unbiased estimate of later error.

```python
from sklearn.metrics import mean_squared_error


def fit_forward_mlp(model, X_fit, y_fit, X_valid, y_valid, epochs):
    """Continue one optimizer path; retain the best validation snapshot, never test."""
    train_loss, val_loss = [], []
    best_model, best_loss, best_epoch = None, np.inf, None
    for epoch in range(1, epochs + 1):
        model.partial_fit(X_fit, y_fit)  # one epoch, retaining Adam's optimizer state
        train_loss.append(mean_squared_error(y_fit, model.predict(X_fit)))
        val_loss.append(mean_squared_error(y_valid, model.predict(X_valid)))
        if val_loss[-1] < best_loss:
            best_loss, best_epoch = val_loss[-1], epoch
            best_model = deepcopy(model)  # copy, not another reference to the mutable model
    return best_model, model, best_epoch, np.asarray(train_loss), np.asarray(val_loss)


# The ordered blocks model a forward protocol; this planted population is iid.
X, y, _ = make_panel(T=900, seed=909)
train_end, val_end = 300, 600
scaler_es = StandardScaler().fit(X[:train_end])
X_fit, X_val = scaler_es.transform(X[:train_end]), scaler_es.transform(X[train_end:val_end])
y_fit, y_val = y[:train_end], y[train_end:val_end]
mlp = MLPRegressor(hidden_layer_sizes=(48, 48), solver='adam',
                   random_state=0, alpha=1e-7, learning_rate_init=0.002)
best_checkpoint, terminal_checkpoint, best_epoch, train_loss, val_loss = fit_forward_mlp(
    mlp, X_fit, y_fit, X_val, y_val, epochs=120)
# Open the final 300 rows only after checkpoint selection.
X_test_es, y_test_es = scaler_es.transform(X[val_end:]), y[val_end:]
restored_test_mse = mean_squared_error(y_test_es, best_checkpoint.predict(X_test_es))
terminal_test_mse = mean_squared_error(y_test_es, terminal_checkpoint.predict(X_test_es))
print(f'Validation selected epoch {best_epoch}/120; test MSE restored={restored_test_mse:.3f}, terminal={terminal_test_mse:.3f}')
fig, axes = plt.subplots(1, 2, figsize=(10, 3.6))
epoch_axis = np.arange(1, len(train_loss) + 1)
axes[0].plot(epoch_axis, train_loss, label='training MSE')
axes[0].plot(epoch_axis, val_loss, label='forward-validation MSE')
axes[0].axvline(best_epoch, color='tab:red', ls='--', label=f'restore epoch {best_epoch}')
axes[0].set(xlabel='epoch', ylabel='MSE', title='Validation chooses the snapshot')
axes[0].legend(fontsize=8)
values = [restored_test_mse, terminal_test_mse]
axes[1].bar(['restored', 'terminal'], values, color=['tab:green', 'tab:red'])
for j, value in enumerate(values): axes[1].text(j, value, f'{value:.3f}', ha='center', va='bottom')
axes[1].set(ylabel='test MSE', title='Later rows evaluate both fixed choices')
fig.tight_layout(); save_lecture_figure(fig, 'demo12_early_stopping.png'); plt.show()
```

### Demo 13 — Match the inductive bias to the target

### Let the target decide which description helps

**Question:** Does the useful representation change when the question changes? The same 2,400 binary rows of length 24 have two labels: whether `1,1,0,1` appears, or whether enough ones appear anywhere. Five percent of labels are flipped. The first 1,500 rows fit; the last 900 test.

The shared-filter proxy computes 16 indicators, one for the presence anywhere of each possible four-bit pattern. A logistic classifier learns from those indicators. The summary model receives just two values: count of ones and count of transitions. Transitions use order, so this is not a wholly order-blind baseline. `LogisticRegression.fit` learns classifier coefficients; `score` reports accuracy on the new rows.

**Predict → change → check:** predict which representation wins in each world before running. The proxy is **not a trained convolutional network**, and 16 indicators versus two summaries is not an equal feature budget. The deliberately favorable constructions test a representation mechanism, not a general ranking of sequence models.

```python
g = np.random.default_rng(1313)
n_rows, length = 2400, 24
sequences = g.integers(0, 2, size=(n_rows, length))
motifs = np.array([
    [(code >> bit) & 1 for bit in range(3, -1, -1)] for code in range(16)
])

def shared_motif_features(rows):
    """One shared four-step filter bank, scanned across every sequence."""
    out = np.zeros((len(rows), len(motifs)))
    for i, row in enumerate(rows):
        windows = np.lib.stride_tricks.sliding_window_view(row, 4)
        out[i] = (windows[:, None, :] == motifs[None, :, :]).all(axis=2).any(axis=0)
    return out

cut = 1500
order_features = shared_motif_features(sequences)
summary_features = np.column_stack([
    sequences.sum(axis=1),
    np.abs(np.diff(sequences, axis=1)).sum(axis=1),
])

# Two known-truth worlds use the same sequences. One target depends on a local
# ordered motif; the other depends only on the row's total number of ones.
targets = {
    "ordered motif": order_features[:, 13].astype(int),
    "order-free count": (sequences.sum(axis=1) >= 12).astype(int),
}
scores = []
for world, clean_target in targets.items():
    noisy_target = clean_target.copy()
    world_rng = np.random.default_rng(1400 if world == "ordered motif" else 1500)
    flip = world_rng.random(n_rows) < 0.05
    noisy_target = np.where(flip, 1 - noisy_target, noisy_target)
    models = {
        "shared-filter proxy": (order_features, LogisticRegression(max_iter=1000)),
        "summary baseline": (summary_features, LogisticRegression(max_iter=1000)),
    }
    for model_name, (features, model) in models.items():
        model.fit(features[:cut], noisy_target[:cut])
        accuracy = np.mean(model.predict(features[cut:]) == noisy_target[cut:])
        scores.append({"world": world, "model": model_name, "accuracy": accuracy})

sequence_results = pd.DataFrame(scores)
print(sequence_results.pivot(index="world", columns="model", values="accuracy").round(3))

fig, ax = plt.subplots(figsize=(8, 3.6))
worlds = list(targets)
x = np.arange(len(worlds)); width = .34
for offset, (model_name, color) in zip(
    [-width/2, width/2],
    [("shared-filter proxy", "tab:blue"), ("summary baseline", "0.55")],
):
    values = [
        sequence_results.query("world == @world and model == @model_name")["accuracy"].iloc[0]
        for world in worlds
    ]
    ax.bar(x + offset, values, width, label=model_name, color=color)
ax.set_xticks(x, ["ordered-motif target", "order-free count target"])
ax.set_ylim(.45, 1)
ax.set_ylabel("held-out accuracy")
ax.set_title("An inductive bias helps only when it matches the target")
ax.legend(frameon=False)
plt.tight_layout()
save_lecture_figure(fig, "demo13_sequence_order.png")
plt.show()
```

### Carry every forecast through the same ledger

**Question:** Does better prediction still help after choosing positions and paying to trade? The simulated panel has 520 dates, 30 assets, and 10 persistent features. `panel_y[day]` is the subsequent return associated with that day's observed features. All models fit the first 240 dates; days 240–299 select the MLP checkpoint; dates 300–319 are a gap; the final 200 dates test. Input scaling and MLP target scaling use only the training block.

Ridge, boosting, and the restored MLP produce one forecast per test asset-date. `cross_sectional_weights` centers ranks or clipped prediction magnitudes **after** the transformation, then divides by total absolute weight. Both targets have net zero and gross one; tied/constant forecasts are handled without inventing positions. `costed_panel_policy` reuses Demo 3's actual-holdings ledger with no inaction threshold and a 2 bp charge per full-L1 dollar traded, including entry.

The panels connect daily rank IC, net Sharpe under the two sizing rules, and actual traded turnover. Maximum drawdown includes the initial dollar of wealth. Financing, market impact, and short-borrow charges remain absent. Strong planted effects and diversification produce large Sharpes; these are mechanism results, not plausible market performance estimates.

**Predict → change → check:** increase only the fee or only the sizing rule while preserving all forecasts. Reconcile the first two days' holdings, return, fee, and new wealth by hand. If you choose a model or policy after seeing these test scores, that choice requires a new final test.

```python
panel_rng = np.random.default_rng(1414)
n_days, n_assets, n_features = 520, 30, 10
panel_x = np.zeros((n_days, n_assets, n_features))
panel_x[0] = panel_rng.standard_normal((n_assets, n_features))
for day in range(1, n_days):
    panel_x[day] = (
        0.65 * panel_x[day - 1]
        + np.sqrt(1 - 0.65 ** 2) * panel_rng.standard_normal((n_assets, n_features))
    )

panel_truth = 0.0015 * (
    ((panel_x[:, :, 0] > 0.4) & (panel_x[:, :, 1] < -0.2)).astype(float)
    - ((panel_x[:, :, 2] < -0.5) & (panel_x[:, :, 3] > 0.0)).astype(float)
) + 0.0007 * panel_x[:, :, 4]
panel_y = panel_truth + 0.012 * panel_rng.standard_normal((n_days, n_assets))
train_days, validation_end, test_start = 240, 300, 320
X_train = panel_x[:train_days].reshape(-1, n_features)
y_train = panel_y[:train_days].ravel()
X_valid_panel = panel_x[train_days:validation_end].reshape(-1, n_features)
y_valid_panel = panel_y[train_days:validation_end].ravel()
X_test = panel_x[test_start:].reshape(-1, n_features)

panel_scaler = StandardScaler().fit(X_train)
panel_models = {
    "Ridge": Ridge(alpha=10.0),
    "Boosting": GradientBoostingRegressor(
        n_estimators=100, max_depth=2, learning_rate=0.05, random_state=14
    ),
    "MLP": MLPRegressor(
        hidden_layer_sizes=(16, 16), alpha=0.001, random_state=14,
    ),
}
panel_predictions = {}
for model_name, model in panel_models.items():
    if model_name == 'MLP':
        target_mean, target_scale = y_train.mean(), y_train.std()
        model, _, panel_best_epoch, _, _ = fit_forward_mlp(
            model, panel_scaler.transform(X_train), (y_train - target_mean) / target_scale,
            panel_scaler.transform(X_valid_panel), (y_valid_panel - target_mean) / target_scale,
            epochs=250)
        pred = model.predict(panel_scaler.transform(X_test)) * target_scale + target_mean
        print(f'Demo 14 MLP checkpoint: {panel_best_epoch}; selected only on days 240–299')
    else:
        model.fit(panel_scaler.transform(X_train), y_train)
        pred = model.predict(panel_scaler.transform(X_test))
    panel_predictions[model_name] = pred.reshape(n_days - test_start, n_assets)


def cross_sectional_weights(prediction, rule):
    if rule == "rank":
        raw = pd.DataFrame(prediction).rank(axis=1).to_numpy()
        raw -= raw.mean(axis=1, keepdims=True)
    elif rule == "magnitude":
        centered = prediction - prediction.mean(axis=1, keepdims=True)
        scale = np.quantile(np.abs(centered), 0.90, axis=1, keepdims=True)
        raw = np.clip(centered, -scale, scale)
    else:
        raise ValueError(rule)
    raw -= raw.mean(axis=1, keepdims=True)  # clipping can also change the mean
    gross = np.abs(raw).sum(axis=1, keepdims=True)
    return np.divide(raw, gross, out=np.zeros_like(raw), where=gross > 0)


def costed_panel_policy(weights, returns, cost_bps=2.0):
    result = execute_policy(weights, returns, cost_bps=cost_bps)
    return result['gross'], result['net'], result['turnover']


panel_rows = []
test_y = panel_y[test_start:]  # y[day] is the subsequent return for features observed that day
test_truth = panel_truth[test_start:]
for model_name, prediction in panel_predictions.items():
    prediction_mse = np.mean((prediction - test_truth) ** 2)
    daily_ic = np.mean([
        met.rank_ic(prediction[day], test_y[day])
        for day in range(len(prediction))
    ])
    for sizing_rule in ("rank", "magnitude"):
        weights = cross_sectional_weights(prediction, sizing_rule)
        gross, net, turnover = costed_panel_policy(weights, test_y, cost_bps=2.0)
        wealth = np.cumprod(1 + net)
        drawdown = wealth / np.maximum.accumulate(np.r_[1.0, wealth])[1:] - 1
        panel_rows.append({
            "model": model_name,
            "sizing": sizing_rule,
            "truth_RMSE_bp": np.sqrt(prediction_mse) * 1e4,
            "rank_IC": daily_ic,
            "turnover": turnover.mean(),
            "gross_Sharpe": np.sqrt(252) * gross.mean() / gross.std(ddof=1),
            "net_Sharpe": np.sqrt(252) * net.mean() / net.std(ddof=1),
            "max_drawdown": drawdown.min(),
        })

end_to_end = pd.DataFrame(panel_rows)
print("One test panel, common clock and 2 bp linear cost per L1 dollar traded:")
print(end_to_end.round(3).to_string(index=False))

fig, axes = plt.subplots(1, 3, figsize=(11, 3.6))
model_order = ["Ridge", "Boosting", "MLP"]
prediction_summary = end_to_end.drop_duplicates("model").set_index("model")
axes[0].bar(model_order, prediction_summary.loc[model_order, "rank_IC"], color="0.45")
axes[0].set_title("Prediction ranking")
axes[0].set_ylabel("mean daily rank IC")
x = np.arange(len(model_order)); width = .34
for offset, sizing_rule, color in [
    (-width / 2, "rank", "tab:blue"),
    (width / 2, "magnitude", "tab:orange"),
]:
    values = end_to_end.query("sizing == @sizing_rule").set_index("model")
    axes[1].bar(x + offset, values.loc[model_order, "net_Sharpe"], width,
                label=sizing_rule, color=color)
    axes[2].bar(x + offset, values.loc[model_order, "turnover"], width,
                label=sizing_rule, color=color)
axes[1].set_title("Same forecasts, different sizing")
axes[1].set_ylabel("net Sharpe after 2 bp")
axes[2].set_title("Implementation changes too")
axes[2].set_ylabel("mean daily L1 turnover")
for ax in axes[1:]:
    ax.set_xticks(x, model_order)
axes[1].legend(frameon=False)
fig.suptitle("Prediction quality does not uniquely determine portfolio value")
fig.tight_layout(rect=(0, 0, 1, .91))
save_lecture_figure(fig, "demo14_forecast_to_portfolio.png")
plt.show()
# Expected: the forecast ranking is only the first verdict. Model rankings can
#           change after sizing, turnover, and costs, and can differ by sizing rule.
```

## Where this leaves us

A forecast has economic meaning only after it passes through a decision rule, risk constraints, and implementation costs. A neural representation likewise earns value only by improving later decisions beyond simpler models on the same information.
