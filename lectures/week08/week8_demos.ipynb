### Demo 1 — Three sizing schemes from one row of scores

> **Break cue:** Deck A, after recording segment S1. Read the task on the preceding slide, predict what the output should show, then run the cell. Record what the task asks for: the specific values you observed, the sample and settings that produced them, and one conclusion the result cannot support.

```python
K = 60
scores = rng.standard_normal(K)
ranks = np.argsort(np.argsort(scores))

def normalize_dn(w):
    """Dollar-neutral, gross exposure 1."""
    w = w - w.mean()
    return w / np.abs(w).sum()

# Quantile L/S — top quintile +, bottom quintile -, else 0
q = K // 5
order = np.argsort(scores)
w_quant = np.zeros(K)
w_quant[order[-q:]] = 1.0; w_quant[order[:q]] = -1.0
w_quant = normalize_dn(w_quant)

# Rank-proportional
w_rank = normalize_dn(ranks - ranks.mean())

# Signal-scaled (z-scored)
z = (scores - scores.mean()) / scores.std()
w_sig = normalize_dn(z)

fig, ax = plt.subplots(figsize=(8, 3.5))
idx = np.argsort(scores)
ax.plot(scores[idx], w_quant[idx], "o-", label="quantile L/S", ms=3)
ax.plot(scores[idx], w_rank[idx], "s-", label="rank-prop", ms=3)
ax.plot(scores[idx], w_sig[idx], "^-", label="signal-scaled", ms=3)
ax.set_xlabel("score"); ax.set_ylabel("weight")
ax.set_title("Three sizing schemes — same forecast, three position profiles")
ax.legend(); ax.axhline(0, lw=0.5, color="k"); plt.show()
# Expected: one row of scores produces three different weight vectors. Sizing is a
#           decision, not a rescaling.
```

### Demo 2 — Volatility targeting on a regime-switching market

> **Break cue:** Deck A, after recording segment S1. Read the task on the preceding slide, predict what the output should show, then run the cell. Record what the task asks for: the specific values you observed, the sample and settings that produced them, and one conclusion the result cannot support.

```python
r = np.asarray(sim.regime_switching(
    n=2000, sig_calm=0.006, sig_turb=0.022,
    p_stay_calm=0.985, p_stay_turb=0.96, seed=11
))

target_vol = 0.10 / np.sqrt(252)   # 10% annualized target
trailing = pd.Series(r).rolling(20).std().shift(1)   # past-only
lev = (target_vol / trailing).clip(0, 5).to_numpy()
lev = np.nan_to_num(lev, nan=1.0)

pnl_const = r
pnl_vt = lev * r

print(f"Constant Sharpe: {met.sharpe(pnl_const):+.2f}, MDD: {met.max_drawdown(pnl_const):.2%}")
print(f"Vol-targeted   : {met.sharpe(pnl_vt):+.2f}, MDD: {met.max_drawdown(pnl_vt):.2%}")

fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].plot(np.cumsum(pnl_const), label="constant size")
axes[0].plot(np.cumsum(pnl_vt), label="vol targeted")
axes[0].set_title("Cumulative P&L"); axes[0].legend()
axes[1].plot(trailing, lw=0.5, color="grey", label="trailing vol")
axes[1].plot(lev, lw=0.5, color="tab:blue", label="leverage")
axes[1].set_title("Leverage mirrors vol"); axes[1].legend()
plt.show()
```

### Demo 3 — Select a no-trade band, then freeze it

> **Break cue:** Deck A, after recording segment S1. Read the task on the preceding slide, predict what the output should show, then run the cell. Record what the task asks for: the specific values you observed, the sample and settings that produced them, and one conclusion the result cannot support.

```python
T, K = 2200, 30
score_path = rng.standard_normal((T, K)) * 0.5 + rng.standard_normal((1, K)) * 0.3
fwd_ret = rng.standard_normal((T, K)) * 0.01
fwd_ret[1:] += 0.0005 * score_path[:-1]  # score at t predicts return at t+1

def weights_from_score(s):
    s = s - s.mean(1, keepdims=True)
    return s / np.abs(s).sum(1, keepdims=True)

w_raw = weights_from_score(score_path)
# EWMA smoothed
alpha = 0.2
s_smooth = np.zeros_like(score_path); s_smooth[0] = score_path[0]
for t in range(1, T):
    s_smooth[t] = alpha * score_path[t] + (1 - alpha) * s_smooth[t-1]
w_smooth = weights_from_score(s_smooth)

# Per-period P&L
def strat_pnl(w, ret, cost_bps):
    g = (w[:-1] * ret[1:]).sum(1)
    turn = np.abs(w[1:] - w[:-1]).sum(1)
    return g - turn * cost_bps / 1e4

bps_grid = np.linspace(0, 8, 17)
net_raw = [met.sharpe(strat_pnl(w_raw, fwd_ret, c)) for c in bps_grid]
net_smooth = [met.sharpe(strat_pnl(w_smooth, fwd_ret, c)) for c in bps_grid]
print(f"Turnover raw    : {np.abs(np.diff(w_raw, axis=0)).sum(1).mean():.2f}/day")
print(f"Turnover smooth : {np.abs(np.diff(w_smooth, axis=0)).sum(1).mean():.2f}/day")

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(bps_grid, net_raw, "o-", label="raw")
ax.plot(bps_grid, net_smooth, "s-", label="smoothed")
ax.axhline(0, color="k", lw=0.5)
ax.set_xlabel("cost (bps per unit turnover)"); ax.set_ylabel("net Sharpe")
ax.set_title("The inversion: slower, weaker signal wins after costs")
ax.legend(); fig.tight_layout()
save_lecture_figure(fig, "demo3_cost_inversion.png")
plt.show()

# The break task uses a validation block to choose the no-trade band and an
# untouched test block to evaluate the frozen choice.
def band_positions(target_weights, half_width):
    actual = np.zeros_like(target_weights)
    actual[0] = target_weights[0]
    for t in range(1, len(target_weights)):
        trade = target_weights[t] - actual[t - 1]
        actual[t] = np.where(np.abs(trade) > half_width,
                             target_weights[t], actual[t - 1])
    return actual

band_rows = []
band_paths = {}
policy_cut = 1100
for band in [0.0, 0.005, 0.010, 0.020, 0.030]:
    actual = band_positions(w_raw, band)
    pnl = strat_pnl(actual, fwd_ret, cost_bps=2.0)
    turnover_path = np.abs(np.diff(actual, axis=0)).sum(1)
    band_paths[band] = actual
    band_rows.append({
        "half_width": band,
        "validation_net_Sharpe": met.sharpe(pnl[:policy_cut - 1]),
        "test_net_Sharpe": met.sharpe(pnl[policy_cut - 1:]),
        "test_tracking_error": np.mean(np.abs(actual[policy_cut:] - w_raw[policy_cut:])),
        "test_turnover": turnover_path[policy_cut - 1:].mean(),
    })
band_frame = pd.DataFrame(band_rows)
selected_band = float(
    band_frame.loc[band_frame["validation_net_Sharpe"].idxmax(), "half_width"]
)
print("\nValidation selects the no-trade band; test evaluates the frozen choice:")
print(band_frame.round(3).to_string(index=False))
print(f"Selected half-width: {selected_band:.3f}")
fig, axes = plt.subplots(1, 4, figsize=(13, 3.4))
for ax, column, title in zip(
    axes,
    ["validation_net_Sharpe", "test_net_Sharpe", "test_turnover", "test_tracking_error"],
    ["Select on validation", "Evaluate on test", "Test turnover", "Test tracking error"],
):
    ax.plot(band_frame["half_width"], band_frame[column], "o-", lw=2)
    chosen = band_frame[band_frame["half_width"] == selected_band].iloc[0]
    ax.scatter([selected_band], [chosen[column]], s=75, marker="*", color="tab:red", zorder=4)
    ax.set(xlabel="no-trade half-width", title=title)
    ax.grid(alpha=.2)
axes[0].set_ylabel("net Sharpe at 2 bps")
axes[1].set_ylabel("net Sharpe at 2 bps")
axes[2].set_ylabel("turnover per day")
axes[3].set_ylabel("mean absolute error")
fig.suptitle("Demo 3: choose the implementation policy before opening the test block")
fig.tight_layout(rect=(0, 0, 1, .91))
save_lecture_figure(fig, "demo3_no_trade_bands.png")
plt.show()
# Expected: validation selects one band; its test result is reported without
#           re-selecting on the test block.
```

### Guided bridge — two nearly redundant assets expose plug-in instability

> **Demonstration:** State what you expect before running the cell; afterward, explain which output supports or contradicts that expectation.

```python
pair_rng = np.random.default_rng(808)
pair_dates = pd.bdate_range("2024-01-02", periods=520)
pair_common = pair_rng.normal(0.00025, 0.009, len(pair_dates))
pair_returns = pd.DataFrame(
    {
        "Asset A": pair_common + pair_rng.normal(0, 0.0013, len(pair_dates)),
        "Asset B": pair_common + pair_rng.normal(0, 0.0013, len(pair_dates)),
    },
    index=pair_dates,
)
pair_window = 60
pair_gamma = 100.0
pair_rows = []
for endpoint in range(pair_window, len(pair_returns)):
    history = pair_returns.iloc[endpoint - pair_window:endpoint]
    mean = history.mean().to_numpy() * 252
    covariance = history.cov().to_numpy() * 252
    inverse = np.linalg.inv(covariance)
    ones = np.ones(2)
    # Fully invested unconstrained quadratic-utility solution. The Lagrange
    # multiplier imposes w_A + w_B = 1 while permitting the long/short pair.
    lagrange = (ones @ inverse @ mean - pair_gamma) / (ones @ inverse @ ones)
    unconstrained = inverse @ (mean - lagrange * ones) / pair_gamma
    inverse_vol = 1 / np.sqrt(np.diag(covariance))
    inverse_vol /= inverse_vol.sum()
    pair_rows.append(
        {
            "date": pair_returns.index[endpoint],
            "correlation": history.corr().iloc[0, 1],
            "mean spread": mean[0] - mean[1],
            "plug-in": unconstrained[0],
            "long-only plug-in": np.clip(unconstrained[0], 0, 1),
            "equal weight": 0.5,
            "inverse volatility": inverse_vol[0],
        }
    )
pair_rolling = pd.DataFrame(pair_rows).set_index("date")

fig, axes = plt.subplots(1, 3, figsize=(11, 3.2))
(1 + pair_returns).cumprod().plot(ax=axes[0], lw=1.4)
axes[0].set_title("The return paths look redundant")
axes[0].set_ylabel("growth of $1")
axes[0].legend(frameon=False, fontsize=8)
axes[1].plot(pair_rolling.index, pair_rolling["correlation"], color="#3b6fb6")
axes[1].axhline(pair_rolling["correlation"].median(), color="0.25", ls=":", lw=1)
axes[1].set_ylim(0.94, 1.001)
axes[1].set_title("Rolling correlation stays near one")
axes[1].set_ylabel("60-day correlation")
axes[2].plot(pair_rolling.index, pair_rolling["plug-in"], color="#b23a48", lw=1.2)
axes[2].axhline(0.5, color="0.25", ls=":", lw=1)
axes[2].set_title("Yet the optimized A weight swings")
axes[2].set_ylabel("fully invested weight in A")
for ax in axes:
    ax.grid(alpha=0.18)
for ax in axes[1:]:
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
fig.tight_layout()
save_lecture_figure(fig, "pair_mean_variance_instability.png")
plt.show()

method_columns = ["long-only plug-in", "equal weight", "inverse volatility"]
method_colors = ["#b23a48", "#4c956c", "#d89b2b"]
turnover = {}
fig, axes = plt.subplots(1, 2, figsize=(10, 3.25), gridspec_kw={"width_ratios": [2.2, 1]})
for method, color in zip(method_columns, method_colors):
    axes[0].plot(pair_rolling.index, pair_rolling[method], label=method, color=color, lw=1.35)
    turnover[method] = pair_rolling[method].diff().abs().dropna().mean()
axes[0].set_ylim(-0.04, 1.04)
axes[0].set_title("How much capital goes to Asset A?")
axes[0].set_ylabel("long-only weight")
axes[0].legend(frameon=False, fontsize=8, ncol=3, loc="lower center")
axes[1].bar(range(3), [turnover[name] for name in method_columns], color=method_colors)
axes[1].set_xticks(range(3), ["plug-in", "1/N", "inverse vol"])
axes[1].set_title("Daily target turnover")
axes[1].set_ylabel("mean absolute change")
for index, method in enumerate(method_columns):
    axes[1].text(index, turnover[method] + 0.001, f"{turnover[method]:.3f}", ha="center", fontsize=8)
for ax in axes:
    ax.grid(alpha=0.18, axis="y")
fig.tight_layout()
save_lecture_figure(fig, "pair_low_estimation_allocators.png")
plt.show()
print("Median rolling correlation:", round(pair_rolling["correlation"].median(), 3))
print("Unconstrained A-weight range:", tuple(pair_rolling["plug-in"].agg(["min", "max"]).round(2)))
print("Mean daily target turnover:", {name: round(value, 4) for name, value in turnover.items()})
# Expected: two nearly redundant return paths generate a volatile plug-in
# allocation. Equal weight ignores the noisy distinction; inverse volatility
# estimates only marginal risks, so both are much more stable in this example.
```

### Demo 4 — Mean-variance error maximization

> **Break cue:** Deck A, after recording segment S2. Read the task on the preceding slide, predict what the output should show, then run the cell. Record what the task asks for: the specific values you observed, the sample and settings that produced them, and one conclusion the result cannot support.

```python
K = 12
rho = 0.84
Sigma = rho * np.ones((K, K)) + (1 - rho) * np.eye(K)
mu = rng.standard_normal(K) * 0.02
base = np.linalg.solve(Sigma, mu)
perts = [np.linalg.solve(Sigma, mu + 0.25 * mu.std() * rng.standard_normal(K)) for _ in range(4)]

fig, ax = plt.subplots(figsize=(8, 3.5))
x = np.arange(K)
ax.bar(x - 0.30, base, width=0.15, label="base")
for i, p in enumerate(perts):
    ax.bar(x + (i - 1) * 0.15, p, width=0.15, label=f"pert {i+1}")
ax.axhline(0, color="k", lw=0.5)
ax.set_title("Mean-variance weights move under 1/4-σ input perturbations")
ax.set_xlabel("asset")
ax.set_ylabel("unconstrained weight direction")
ax.legend(ncol=5, fontsize=8); fig.tight_layout()
save_lecture_figure(fig, "demo4_mean_variance_perturbations.png")
plt.show()
flips = np.mean([(np.sign(p) != np.sign(base)).mean() for p in perts])
print(f"Avg fraction of sign flips per perturbation: {flips:.0%}")
# Expected: ~4% of position signs flip under small input perturbations. Mean-variance
#           maximizes estimation error.
```

> **Break cue:** Deck A, after recording segment S2. Read the task on the preceding slide, predict what the output should show, then run the cell. Record what the task asks for: the specific values you observed, the sample and settings that produced them, and one conclusion the result cannot support.

```python
def long_only_gmv(covariance):
    """Long-only, fully invested global-minimum-variance portfolio."""
    n_assets = covariance.shape[0]
    result = minimize(
        lambda weight: float(weight @ covariance @ weight),
        np.repeat(1 / n_assets, n_assets),
        jac=lambda weight: 2 * covariance @ weight,
        method="SLSQP",
        bounds=[(0.0, 1.0)] * n_assets,
        constraints={
            "type": "eq",
            "fun": lambda weight: weight.sum() - 1.0,
            "jac": lambda weight: np.ones_like(weight),
        },
        options={"ftol": 1e-10, "maxiter": 250},
    )
    if not result.success:
        raise RuntimeError(result.message)
    return result.x


def one_sample(K=20, T=120, seed=None):
    g = np.random.default_rng(seed)
    vol = np.linspace(0.012, 0.026, K)
    loadings = np.column_stack([
        np.linspace(0.2, 1.0, K),
        np.sin(np.linspace(0, 2 * np.pi, K)),
    ])
    factor_cov = np.diag([0.00010, 0.00004])
    cov = loadings @ factor_cov @ loadings.T + np.diag(0.45 * vol ** 2)
    returns = g.multivariate_normal(np.zeros(K), cov, size=T)
    sample_cov = np.cov(returns, rowvar=False)
    shrink_cov = LedoitWolf().fit(returns).covariance_

    oracle = long_only_gmv(cov)
    oracle_variance = float(oracle @ cov @ oracle)
    w_eq = np.ones(K) / K
    w_iv = 1 / np.sqrt(np.diag(sample_cov)); w_iv /= w_iv.sum()
    weights = {
        "Equal weight": w_eq,
        "Inverse vol": w_iv,
        "Sample GMV": long_only_gmv(sample_cov),
        "LW GMV": long_only_gmv(shrink_cov),
    }
    return {
        name: float(weight @ cov @ weight) / oracle_variance
        for name, weight in weights.items()
    }

results = [one_sample(seed=i) for i in range(120)]
df = pd.DataFrame(results)
print("Median population variance relative to oracle GMV:")
print(df.median().round(3))

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.boxplot(
    [df[c].dropna() for c in df.columns],
    tick_labels=df.columns,
    showmeans=True,
)
ax.set_title("One mandate: long-only, fully invested, minimum variance")
ax.axhline(1, color="k", lw=0.8, ls=":", label="oracle GMV")
ax.set_ylabel("population variance / oracle variance")
ax.legend(frameon=False)
fig.tight_layout()
save_lecture_figure(fig, "demo5_robust_allocations.png")
plt.show()
# Expected: the sample optimizer is fragile with short histories; shrinkage
#           improves it without changing the mandate.
```

> **Optional empirical comparison:** Deck A, after recording segment S2. Run this after the known-truth demo. Use the same transformation and compare the simulation with the historical series. Record one important difference and the most likely reason for it.

```python
portfolio_raw, portfolio_source = course_csv("datasets/famafrench/ff_12industry_daily.csv")
portfolio_raw["date"] = pd.to_datetime(portfolio_raw["date"].astype(str), format="%Y%m%d")
portfolio_returns = portfolio_raw.set_index("date").sort_index().loc["2000":] / 100.0
portfolio_values = portfolio_returns.to_numpy()
portfolio_names = portfolio_returns.columns

def minimum_variance_weights(covariance):
    ones = np.ones(covariance.shape[0])
    raw = np.linalg.solve(covariance + 1e-12 * np.eye(len(ones)), ones)
    return raw / raw.sum()

lookback, holding = 60, 20
portfolio_paths = {"equal weight": [], "sample covariance": [], "Ledoit-Wolf": []}
portfolio_weight_paths = {name: [] for name in portfolio_paths}
portfolio_dates = []
sample_conditions, shrink_conditions = [], []
for start in range(lookback, len(portfolio_values) - holding + 1, holding):
    estimation = portfolio_values[start - lookback:start]
    realized_block = portfolio_values[start:start + holding]
    sample_covariance = np.cov(estimation, rowvar=False)
    shrink_covariance = LedoitWolf().fit(estimation).covariance_
    weights = {
        "equal weight": np.repeat(1 / len(portfolio_names), len(portfolio_names)),
        "sample covariance": minimum_variance_weights(sample_covariance),
        "Ledoit-Wolf": minimum_variance_weights(shrink_covariance),
    }
    for name, weight in weights.items():
        portfolio_paths[name].extend(realized_block @ weight)
        portfolio_weight_paths[name].append(weight)
    portfolio_dates.extend(portfolio_returns.index[start:start + holding])
    sample_conditions.append(np.linalg.cond(sample_covariance))
    shrink_conditions.append(np.linalg.cond(shrink_covariance))

portfolio_summary = []
for name, values in portfolio_paths.items():
    values = np.asarray(values)
    weight_array = np.asarray(portfolio_weight_paths[name])
    turnover = np.abs(np.diff(weight_array, axis=0)).sum(axis=1).mean() if len(weight_array) > 1 else 0.0
    portfolio_summary.append({
        "portfolio": name,
        "ann_vol": values.std(ddof=1) * np.sqrt(252),
        "ann_Sharpe": values.mean() / values.std(ddof=1) * np.sqrt(252),
        "mean_max_abs_weight": np.abs(weight_array).max(axis=1).mean(),
        "rebalance_turnover": turnover,
    })

print("Source: Kenneth R. French Data Library, bundled 12 industry portfolios")
print(f"File: {portfolio_source}")
print(f"Evaluation: {portfolio_dates[0].date()} through {portfolio_dates[-1].date()}")
print("Units: decimal daily returns; 60-session trailing estimate; weights held for 20 sessions")
print(f"Median covariance condition number: sample={np.median(sample_conditions):.1f}, "
      f"Ledoit-Wolf={np.median(shrink_conditions):.1f}")
print(pd.DataFrame(portfolio_summary).round(3).to_string(index=False))

fig, ax = plt.subplots(figsize=(8, 3.5))
for name, values in portfolio_paths.items():
    ax.plot(portfolio_dates, np.cumsum(values), label=name, lw=0.9)
ax.set_title("Frozen rolling portfolios on subsequent industry returns")
ax.set_ylabel("Cumulative arithmetic return")
ax.legend(); fig.tight_layout(); plt.show()
print("This sample compares estimators under one schedule; it does not establish a universal portfolio ranking.")
```

### Demo 7 — Audit backprop with finite differences

> **Break cue:** Deck B, after recording segment S4. Read the task on the preceding slide, predict what the output should show, then run the cell. Record what the task asks for: the specific values you observed, the sample and settings that produced them, and one conclusion the result cannot support.

```python
parameters = np.array([0.8, -0.3, 1.2, 0.1])  # w1, b1, w2, b2
x_value, target_value = 0.7, -0.4

def scalar_loss_and_gradient(theta):
    w1, b1, w2, b2 = theta
    hidden = np.tanh(w1 * x_value + b1)
    prediction = w2 * hidden + b2
    error = prediction - target_value
    loss = 0.5 * error ** 2
    hidden_grad = error * w2 * (1 - hidden ** 2)
    gradient = np.array([hidden_grad * x_value, hidden_grad, error * hidden, error])
    return loss, gradient

loss, analytic = scalar_loss_and_gradient(parameters)
steps = np.logspace(-14, -1, 28)
errors = []
for step in steps:
    numeric = np.zeros_like(parameters)
    for j in range(len(parameters)):
        up, down = parameters.copy(), parameters.copy()
        up[j] += step
        down[j] -= step
        numeric[j] = (
            scalar_loss_and_gradient(up)[0] - scalar_loss_and_gradient(down)[0]
        ) / (2 * step)
    errors.append(np.max(np.abs(numeric - analytic)))

best = int(np.argmin(errors))
print("Analytic gradient:", np.round(analytic, 6))
print(f"Best finite-difference step={steps[best]:.1e}; max error={errors[best]:.2e}")
print("Very small steps lose precision; large steps cease to be local.")

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.loglog(steps, errors, "o-")
ax.axvline(steps[best], color="tab:red", ls="--", lw=1)
ax.set_xlabel("central-difference step")
ax.set_ylabel("maximum absolute gradient error")
ax.set_title("A gradient check fails at both numerical extremes")
plt.tight_layout()
save_lecture_figure(fig, "demo7_gradient_check.png")
plt.show()
```

### Demo 10 — Capacity is relative to the data budget

> **Break cue:** Deck B, after recording segment S5. Read the task on the preceding slide, predict what the output should show, then run the cell. Record what the task asks for: the specific values you observed, the sample and settings that produced them, and one conclusion the result cannot support.

```python
widths = [4, 12, 32]
sample_sizes = [40, 120, 300]
x_grid = np.linspace(-2, 2, 600)
x_shift = np.r_[np.linspace(-2.7, -2.05, 200), np.linspace(2.05, 2.7, 200)]
n_repeats = 10
train_mse = np.zeros((len(sample_sizes), len(widths), n_repeats))
truth_mse = np.zeros_like(train_mse)
shift_mse = np.zeros_like(train_mse)

# This is a one-hidden-layer tanh network with frozen hidden weights and a
# fitted linear readout. The construction isolates width and data budget from
# optimizer luck while remaining a genuine nonlinear network.
for i, n in enumerate(sample_sizes):
    for repeat in range(n_repeats):
        g = np.random.default_rng(808 + 100 * i + repeat)
        x_train = np.sort(g.uniform(-2, 2, n))
        y_train = target(x_train) + 0.30 * g.standard_normal(n)
        for j, width in enumerate(widths):
            hidden_rng = np.random.default_rng(700 + j)
            hidden_w = hidden_rng.normal(0, 2, width)
            hidden_b = hidden_rng.uniform(-2, 2, width)

            def design(values):
                hidden = np.tanh(values[:, None] * hidden_w + hidden_b)
                return np.column_stack([np.ones(len(values)), hidden])

            phi_train = design(x_train)
            penalty = np.diag(np.r_[0.0, np.full(width, 1e-8)])
            readout = np.linalg.solve(
                phi_train.T @ phi_train + penalty, phi_train.T @ y_train
            )
            train_mse[i, j, repeat] = np.mean((y_train - phi_train @ readout) ** 2)
            truth_mse[i, j, repeat] = np.mean((target(x_grid) - design(x_grid) @ readout) ** 2)
            shift_mse[i, j, repeat] = np.mean((target(x_shift) - design(x_shift) @ readout) ** 2)

surfaces = [np.median(values, axis=2) for values in (train_mse, truth_mse, shift_mse)]
print("Median MSE against known truth over ten held-out samples:")
print(pd.DataFrame(surfaces[1], index=sample_sizes, columns=widths).round(3))

fig, axes = plt.subplots(1, 3, figsize=(12, 3.7))
for ax, values, title in zip(
    axes, surfaces,
    ["Fit to noisy labels", "Fresh in-domain truth", "Shifted-domain truth"],
):
    im = ax.imshow(values, cmap="viridis_r", aspect="auto")
    ax.set_xticks(range(len(widths)), widths)
    ax.set_yticks(range(len(sample_sizes)), sample_sizes)
    ax.set_xlabel("hidden width")
    ax.set_ylabel("training observations")
    ax.set_title(title)
    for i in range(len(sample_sizes)):
        for j in range(len(widths)):
            ax.text(j, i, f"{values[i, j]:.2f}", ha="center", va="center",
                    color="white" if values[i, j] > np.median(values) else "black")
fig.suptitle("Demo 10: useful capacity depends on data budget and evaluation domain\n"
             "Each panel has its own color scale; cell values are median MSE")
fig.subplots_adjust(top=.78, right=.98, wspace=.34)
save_lecture_figure(fig, "demo10_capacity_data_budget.png")
plt.show()
# Expected: the best width depends on the data budget: width 12 wins at n=40, and
#           width 32 only becomes competitive as n grows.
```

### Demo 11 — MLP vs Boosting vs Linear on low-signal tabular returns

> **Break cue:** Deck B, after recording segment S6. Read the task on the preceding slide, predict what the output should show, then run the cell. Record what the task asks for: the specific values you observed, the sample and settings that produced them, and one conclusion the result cannot support.

```python
def make_panel(T=2000, K=20, seed=0):
    g = np.random.default_rng(seed)
    X = g.standard_normal((T, K))
    # Local thresholds and interactions are a favorable known-truth case for trees.
    signal = (
        0.30 * ((X[:, 0] > 0.4) & (X[:, 1] < -0.2))
        - 0.24 * ((X[:, 2] < -0.5) & (X[:, 3] > 0.0))
        + 0.18 * np.where(X[:, 4] > 0.8, X[:, 5], 0.0)
    )
    y = signal + 0.65 * g.standard_normal(T)
    return X, y, signal

comparison = []
for seed in range(5):
    X, y, truth = make_panel(T=1800, seed=seed)
    ix = int(0.6 * len(y))
    X_tr, X_te = X[:ix], X[ix:]
    y_tr, y_te = y[:ix], y[ix:]
    truth_te = truth[ix:]
    scaler = StandardScaler().fit(X_tr)
    Xs_tr, Xs_te = scaler.transform(X_tr), scaler.transform(X_te)
    models = {
        "Ridge": Ridge(alpha=1.0),
        "Boosting": GradientBoostingRegressor(
            n_estimators=100, max_depth=2, learning_rate=0.05, random_state=seed
        ),
        "MLP": MLPRegressor(
            hidden_layer_sizes=(16, 16), max_iter=250, early_stopping=True,
            validation_fraction=0.2, n_iter_no_change=12, random_state=seed
        ),
    }
    for name, model in models.items():
        model.fit(Xs_tr, y_tr)
        pred = model.predict(Xs_te)
        comparison.append({
            "seed": seed,
            "model": name,
            "truth_MSE": np.mean((pred - truth_te) ** 2),
            "noisy_rank_IC": met.rank_ic(pred, y_te),
        })

comparison = pd.DataFrame(comparison)
summary = comparison.groupby("model")[["truth_MSE", "noisy_rank_IC"]].median()
print("Median over five out-of-sample simulations:")
print(summary.round(3))

fig, axes = plt.subplots(1, 2, figsize=(9.5, 3.6))
order = ["Ridge", "Boosting", "MLP"]
axes[0].bar(order, summary.loc[order, "truth_MSE"])
axes[0].set_title("Known-truth error")
axes[0].set_ylabel("MSE against conditional mean")
axes[1].bar(order, summary.loc[order, "noisy_rank_IC"])
axes[1].set_title("Noisy-label rank IC")
axes[1].axhline(0, color="k", lw=0.5)
fig.suptitle("The same model comparison under two evaluation targets")
plt.tight_layout(rect=(0, 0, 1, .91))
save_lecture_figure(fig, "demo11_tabular_bakeoff.png")
plt.show()
# Expected: boosting recovers the conditional mean better than the MLP (truth MSE ~0.03
#           vs ~0.06) while the noisy rank IC hides most of that gap.
```

### Demo 12 — Early stopping curve

> **Break cue:** Deck B, after recording segment S6. Read the task on the preceding slide, predict what the output should show, then run the cell. Record what the task asks for: the specific values you observed, the sample and settings that produced them, and one conclusion the result cannot support.

```python
from sklearn.metrics import mean_squared_error
# Respect time explicitly: the validation block follows the training block.
X, y, _ = make_panel(T=900, seed=909)
train_end, val_end = 300, 600
X_fit, y_fit = X[:train_end], y[:train_end]
X_val, y_val = X[train_end:val_end], y[train_end:val_end]
scaler_es = StandardScaler().fit(X_fit)
X_fit = scaler_es.transform(X_fit)
X_val = scaler_es.transform(X_val)

mlp = MLPRegressor(
    hidden_layer_sizes=(48, 48), solver="adam", warm_start=True,
    max_iter=1, random_state=0, alpha=1e-7, learning_rate_init=0.002,
)
train_loss, val_loss = [], []
with warnings.catch_warnings():
    warnings.simplefilter("ignore", ConvergenceWarning)
    for _ in range(120):
        mlp.fit(X_fit, y_fit)
        train_loss.append(mean_squared_error(y_fit, mlp.predict(X_fit)))
        val_loss.append(mean_squared_error(y_val, mlp.predict(X_val)))

best_epoch = int(np.argmin(val_loss)) + 1
print(f"Forward-validation minimum at epoch {best_epoch} of {len(val_loss)}")
print(f"Validation MSE at minimum={min(val_loss):.3f}; terminal={val_loss[-1]:.3f}")

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(train_loss, label="training MSE")
ax.plot(val_loss, label="forward-validation MSE")
ax.axvline(best_epoch - 1, color="tab:red", ls="--", lw=1,
           label=f"restore epoch {best_epoch}")
ax.set_xlabel("epoch"); ax.set_ylabel("loss")
ax.set_title("Training keeps improving after validation turns")
ax.legend()
save_lecture_figure(fig, "demo12_early_stopping.png")
plt.show()
```

```python
g = np.random.default_rng(1313)
n_rows, length = 2400, 24
sequences = g.integers(0, 2, size=(n_rows, length))
motifs = np.array([
    [(code >> bit) & 1 for bit in range(3, -1, -1)] for code in range(16)
])

def shared_motif_features(rows):
    """One shared four-step filter bank, scanned across every sequence."""
    out = np.zeros((len(rows), len(motifs)))
    for i, row in enumerate(rows):
        windows = np.lib.stride_tricks.sliding_window_view(row, 4)
        out[i] = (windows[:, None, :] == motifs[None, :, :]).all(axis=2).any(axis=0)
    return out

cut = 1500
order_features = shared_motif_features(sequences)
summary_features = np.column_stack([
    sequences.sum(axis=1),
    np.abs(np.diff(sequences, axis=1)).sum(axis=1),
])

# Two known-truth worlds use the same sequences. One target depends on a local
# ordered motif; the other depends only on the row's total number of ones.
targets = {
    "ordered motif": order_features[:, 13].astype(int),
    "order-free count": (sequences.sum(axis=1) >= 12).astype(int),
}
scores = []
for world, clean_target in targets.items():
    noisy_target = clean_target.copy()
    world_rng = np.random.default_rng(1400 if world == "ordered motif" else 1500)
    flip = world_rng.random(n_rows) < 0.05
    noisy_target = np.where(flip, 1 - noisy_target, noisy_target)
    models = {
        "shared-filter proxy": (order_features, LogisticRegression(max_iter=1000)),
        "summary baseline": (summary_features, LogisticRegression(max_iter=1000)),
    }
    for model_name, (features, model) in models.items():
        model.fit(features[:cut], noisy_target[:cut])
        accuracy = np.mean(model.predict(features[cut:]) == noisy_target[cut:])
        scores.append({"world": world, "model": model_name, "accuracy": accuracy})

sequence_results = pd.DataFrame(scores)
print(sequence_results.pivot(index="world", columns="model", values="accuracy").round(3))

fig, ax = plt.subplots(figsize=(8, 3.6))
worlds = list(targets)
x = np.arange(len(worlds)); width = .34
for offset, (model_name, color) in zip(
    [-width/2, width/2],
    [("shared-filter proxy", "tab:blue"), ("summary baseline", "0.55")],
):
    values = [
        sequence_results.query("world == @world and model == @model_name")["accuracy"].iloc[0]
        for world in worlds
    ]
    ax.bar(x + offset, values, width, label=model_name, color=color)
ax.set_xticks(x, ["ordered-motif target", "order-free count target"])
ax.set_ylim(.45, 1)
ax.set_ylabel("held-out accuracy")
ax.set_title("An inductive bias helps only when it matches the target")
ax.legend(frameon=False)
plt.tight_layout()
save_lecture_figure(fig, "demo13_sequence_order.png")
plt.show()
```

### Demo 14 — Carry the forecast comparison into one portfolio policy

> **Demonstration:** State what you expect before running the cell; afterward, explain which output supports or contradicts that expectation.

```python
panel_rng = np.random.default_rng(1414)
n_days, n_assets, n_features = 520, 30, 10
panel_x = np.zeros((n_days, n_assets, n_features))
panel_x[0] = panel_rng.standard_normal((n_assets, n_features))
for day in range(1, n_days):
    panel_x[day] = (
        0.65 * panel_x[day - 1]
        + np.sqrt(1 - 0.65 ** 2) * panel_rng.standard_normal((n_assets, n_features))
    )

panel_truth = 0.0015 * (
    ((panel_x[:, :, 0] > 0.4) & (panel_x[:, :, 1] < -0.2)).astype(float)
    - ((panel_x[:, :, 2] < -0.5) & (panel_x[:, :, 3] > 0.0)).astype(float)
) + 0.0007 * panel_x[:, :, 4]
panel_y = panel_truth + 0.012 * panel_rng.standard_normal((n_days, n_assets))
train_days, test_start = 300, 320
X_train = panel_x[:train_days].reshape(-1, n_features)
y_train = panel_y[:train_days].ravel()
X_test = panel_x[test_start:].reshape(-1, n_features)

panel_scaler = StandardScaler().fit(X_train)
panel_models = {
    "Ridge": Ridge(alpha=10.0),
    "Boosting": GradientBoostingRegressor(
        n_estimators=100, max_depth=2, learning_rate=0.05, random_state=14
    ),
    "MLP": MLPRegressor(
        hidden_layer_sizes=(16, 16), alpha=0.001, max_iter=250,
        early_stopping=True, validation_fraction=0.15,
        n_iter_no_change=12, random_state=14,
    ),
}
panel_predictions = {}
for model_name, model in panel_models.items():
    model.fit(panel_scaler.transform(X_train), y_train)
    panel_predictions[model_name] = model.predict(
        panel_scaler.transform(X_test)
    ).reshape(n_days - test_start, n_assets)


def cross_sectional_weights(prediction, rule):
    if rule == "rank":
        raw = pd.DataFrame(prediction).rank(axis=1, pct=True).to_numpy() - 0.5
    elif rule == "magnitude":
        centered = prediction - prediction.mean(axis=1, keepdims=True)
        scale = np.quantile(np.abs(centered), 0.90, axis=1, keepdims=True)
        raw = np.clip(centered, -scale, scale)
    else:
        raise ValueError(rule)
    return raw / np.abs(raw).sum(axis=1, keepdims=True)


def costed_panel_policy(weights, returns, cost_bps=2.0):
    gross, turnover, net = [], [], []
    prior_post_return = np.zeros(weights.shape[1])
    for day, target_weight in enumerate(weights):
        traded = np.abs(target_weight - prior_post_return).sum()
        period_return = float(target_weight @ returns[day])
        gross.append(period_return)
        turnover.append(traded)
        net.append(period_return - cost_bps * 1e-4 * traded)
        denominator = 1 + period_return
        prior_post_return = target_weight * (1 + returns[day]) / denominator
    return np.asarray(gross), np.asarray(net), np.asarray(turnover)


panel_rows = []
test_y = panel_y[test_start:]
test_truth = panel_truth[test_start:]
for model_name, prediction in panel_predictions.items():
    prediction_mse = np.mean((prediction - test_truth) ** 2)
    daily_ic = np.mean([
        met.rank_ic(prediction[day], test_y[day])
        for day in range(len(prediction))
    ])
    for sizing_rule in ("rank", "magnitude"):
        weights = cross_sectional_weights(prediction, sizing_rule)
        gross, net, turnover = costed_panel_policy(weights, test_y, cost_bps=2.0)
        wealth = np.cumprod(1 + net)
        drawdown = wealth / np.maximum.accumulate(wealth) - 1
        panel_rows.append({
            "model": model_name,
            "sizing": sizing_rule,
            "truth_RMSE_bp": np.sqrt(prediction_mse) * 1e4,
            "rank_IC": daily_ic,
            "turnover": turnover.mean(),
            "gross_Sharpe": np.sqrt(252) * gross.mean() / gross.std(ddof=1),
            "net_Sharpe": np.sqrt(252) * net.mean() / net.std(ddof=1),
            "max_drawdown": drawdown.min(),
        })

end_to_end = pd.DataFrame(panel_rows)
print("One test panel, common clock and 2 bp linear cost per L1 dollar traded:")
print(end_to_end.round(3).to_string(index=False))

fig, axes = plt.subplots(1, 3, figsize=(11, 3.6))
model_order = ["Ridge", "Boosting", "MLP"]
prediction_summary = end_to_end.drop_duplicates("model").set_index("model")
axes[0].bar(model_order, prediction_summary.loc[model_order, "rank_IC"], color="0.45")
axes[0].set_title("Prediction ranking")
axes[0].set_ylabel("mean daily rank IC")
x = np.arange(len(model_order)); width = .34
for offset, sizing_rule, color in [
    (-width / 2, "rank", "tab:blue"),
    (width / 2, "magnitude", "tab:orange"),
]:
    values = end_to_end.query("sizing == @sizing_rule").set_index("model")
    axes[1].bar(x + offset, values.loc[model_order, "net_Sharpe"], width,
                label=sizing_rule, color=color)
    axes[2].bar(x + offset, values.loc[model_order, "turnover"], width,
                label=sizing_rule, color=color)
axes[1].set_title("Same forecasts, different sizing")
axes[1].set_ylabel("net Sharpe after 2 bp")
axes[2].set_title("Implementation changes too")
axes[2].set_ylabel("mean daily L1 turnover")
for ax in axes[1:]:
    ax.set_xticks(x, model_order)
axes[1].legend(frameon=False)
fig.suptitle("Prediction quality does not uniquely determine portfolio value")
fig.tight_layout(rect=(0, 0, 1, .91))
save_lecture_figure(fig, "demo14_forecast_to_portfolio.png")
plt.show()
# Expected: the forecast ranking is only the first verdict. Model rankings can
#           change after sizing, turnover, and costs, and can differ by sizing rule.
```
